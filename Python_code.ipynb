```python
import matplotlib.pyplot as plt
df["Circulation Active Year"].value_counts().plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)
plt.title("Library Circulation Activity by Year")
plt.ylabel("# of circulations")
plt.xlabel("year")
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)
plt.title("Library Circulation Activity by Month")
plt.ylabel("# of circulations")
plt.xlabel("month")
plt.show()
```

```python
df['Circulation Active Month Cleaned'] = df['Circulation Active Month'].fillna("Not Captured")
df.isna().sum()
month_order = [
    "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December", "Not Captured"]
df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)
plt.title("Library Circulation Activity by Month")
plt.ylabel("# of circulations")
plt.xlabel("month")
plt.show()
```

```python
df['Age Range Cleaned'] = df['Age Range'].fillna("Not Captured")
df.isna().sum()
month_order = [
    "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December", "Not Captured"]
df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="black",
    edgecolor="pink",
    figsize=(10,6),
)
plt.title("Library Circulation Activity by Month")
plt.ylabel("# of circulations")
plt.xlabel("month")
plt.show()
```
